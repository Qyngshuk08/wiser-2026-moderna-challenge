## 5. Scaling study
Edit the sequence list below to whatever lengths you're benchmarking. Log every row — this table is your scaling/quantum-resource-analysis deliverable.

```python
import random, json
random.seed(0)

def random_rna(n):
    return "".join(random.choice("AUCG") for _ in range(n))

results = []
import sys
sys.path.insert(0, ".")
from run_dwave import run

for n in [20, 30, 40, 60, 80, 100]:
    seq = random_rna(n)
    print(f"--- n={n} ---")
    r = run(seq, mode="hybrid")
    results.append(r)

with open("scaling_results.json", "w") as f:
    json.dump(results, f, indent=2)
print(json.dumps(results, indent=2))
```
